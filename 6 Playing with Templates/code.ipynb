Templates

```python
from langchain_core.prompts import PromptTemplate,SystemMessagePromptTemplate,HumanMessagePromptTemplate,ChatPromptTemplate,FewShotChatMessagePromptTemplate
```

```python
template='''
You are a expert AI Engineer Researcher.
Here is the context for you
{context}
You need to provide a model that suits best on this type of data
'''

templateP = PromptTemplate.from_template(template=template)

message = templateP.format(context=["ABC","DEF"])
print(message)
```

```python
template_System = SystemMessagePromptTemplate.from_template(template=template)
message = template_System.format(context=["Dummy"])
print(f"type:{message.type}\n\ncontent:{message.content}")
```

```python
template_System = HumanMessagePromptTemplate.from_template(template=template)
message = template_System.format(context=["Dummy"])
print(f"type:{message.type}\n\ncontent:{message.content}")
```

```python
message_h="Summarize this document in {word_length} words"
message_s="NOTE : You are assistant , you need to follow the word length provided by user and summarize it in language {language}"

template_h = HumanMessagePromptTemplate.from_template(template=message_h)
template_s = SystemMessagePromptTemplate.from_template(template=message_s)

chatTemplate = ChatPromptTemplate([template_s,template_h])
message = chatTemplate.invoke({'word_length':100,'language':"hindi"})
print(message)
```

```python
examples=[{
    'word_length':130,
    'language':'english'
},{
    'word_length':100,
    'language':'hindi'
}]
te = FewShotChatMessagePromptTemplate(examples=examples,example_prompt=chatTemplate)
print(te)
```
